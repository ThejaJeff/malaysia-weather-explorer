```python
yearly_count = (
    sp
    .groupby("year")["temperature"]
    .count()
    .reset_index(name="temperature_count")
)

yearly_count
```

```python
yearly_months = (
    sp
    .groupby("year")["month"]
    .nunique()
    .reset_index(name="months_available")
)

yearly_months
```

```python
yearly_quality = yearly_count.merge(
    yearly_months,
    on="year"
)

yearly_quality
```

```python
yearly_quality.sort_values("months_available")
```

```python
complete_years = yearly_quality[
    yearly_quality["months_available"] >= 11
]["year"]
```

```python
complete_years
```

```python
sp_complete = sp[
    sp["year"].isin(complete_years)
].copy()
```

```python
sp_complete["year"].unique()
```

```python
yearly_temperature_clean = (
    sp_complete
    .groupby("year")["temperature"]
    .mean()
    .reset_index()
)
```

```python
yearly_temperature_clean
```

```python
plt.figure(figsize=(10,5))

plt.plot(
    yearly_temperature_clean["year"],
    yearly_temperature_clean["temperature"],
    marker="o"
)

plt.xlabel("Year")
plt.ylabel("Average Temperature (°C)")
plt.title("Annual Average Temperature in Sepang")
plt.grid(True)
plt.show()
```

```python
from scipy.stats import linregress
```

```python
trend = linregress(
    yearly_temperature_clean["year"],
    yearly_temperature_clean["temperature"]
)
```

```python
print("Slope:", trend.slope)
print("Intercept:", trend.intercept)
print("R-squared:", trend.rvalue ** 2)
print("p-value:", trend.pvalue)
```

```python
trend.slope * 10
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10, 5))

plt.scatter(
    yearly_temperature_clean["year"],
    yearly_temperature_clean["temperature"]
)

plt.plot(
    yearly_temperature_clean["year"],
    trend.intercept + trend.slope * yearly_temperature_clean["year"]
)

plt.xlabel("Year")
plt.ylabel("Average Temperature (°C)")
plt.title("Annual Average Temperature and Linear Treand - Sepang")

plt.grid(True)
plt.show()
```
