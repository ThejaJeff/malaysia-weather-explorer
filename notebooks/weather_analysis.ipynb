```python
yearly_temperature = (
    sp
    .groupby("year")["temperature"]
    .mean()
    .reset_index()
)
```

```python
yearly_temperature.head()
```

```python
yearly_temperature.tail()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,5))

plt.plot(
    yearly_temperature["year"],
    yearly_temperature["temperature"],
    marker="o"
)

plt.xlabel("Year")
plt.ylabel("Average Temperature (°C)")
plt.title("Average Annual Temperature in Sepang (1996-2024)")

plt.grid(True)
plt.show()
```

```python
yearly_count = (
    sp
    .groupby("year")["temperature"]
    .count()
    .reset_index(name="temperature_count")
)

yearly_count
```

```python
yearly_months = (
    sp
    .groupby("year")["month"]
    .nunique()
    .reset_index(name="months_available")
)

yearly_months
```

```python
yearly_quality = yearly_count.merge(
    yearly_months,
    on="year"
)

yearly_quality
```

```python
yearly_quality.sort_values("months_available")
```

```python
complete_years = yearly_quality[
    yearly_quality["months_available"] >= 11
]["year"]
```

```python
complete_years
```

```python
sp_complete = sp[
    sp["year"].isin(complete_years)
].copy()
```

```python
sp_complete["year"].unique()
```

```python
yearly_temperature_clean = (
    sp_complete
    .groupby("year")["temperature"]
    .mean()
    .reset_index()
)
```

```python
yearly_temperature_clean
```

```python
plt.figure(figsize=(10,5))

plt.plot(
    yearly_temperature_clean["year"],
    yearly_temperature_clean["temperature"],
    marker="o"
)

plt.xlabel("Year")
plt.ylabel("Average Temperature (°C)")
plt.title("Annual Average Temperature in Sepang")
plt.grid(True)
plt.show()
```
